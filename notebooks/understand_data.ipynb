```python
ground_truth_map = {}

for _, row in train_gr.iterrows():
    s1_id = row["source1_entity_id"]
    matches = row["matched_entity_ids"]

    if pd.isna(matches) or str(matches).strip() == "":
        ground_truth_map[s1_id] = set()
    else:
        ground_truth_map[s1_id] = set(
            str(matches).split(",")
        )
```

```python
first_s1 = train_gr.iloc[0]["source1_entity_id"]

print("S1:", first_s1)
print("True matches:", ground_truth_map[first_s1])
```

```python
def get_target(s1_id, candidate_id):
    return int(candidate_id in ground_truth_map.get(s1_id, set()))
```

```python
import re
import unicodedata

def normalize_text(text):
    if pd.isna(text):
        return ""
    
    text = str(text)
    
    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)
    
    # Lowercase
    text = text.casefold()
    
    # Replace punctuation with spaces
    text = re.sub(r"[^\w\s]", " ", text)
    
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

```python
examples = [
    "ABC Technologies Pvt. Ltd.",
    "ABC TECHNOLOGIES PRIVATE LIMITED",
    "ABC-Tech Pvt Ltd"
]

for x in examples:
    print(x, "→", normalize_text(x))
```

```python
train1["name_norm"] = train1["business_name"].apply(normalize_text)

train3["name_norm"] = train3["business_name"].apply(normalize_text)
```

```python
train1.head()
```

```python
def make_block_key(row):
    name = row["name_norm"]
    country = str(row["country"]).casefold().strip()
    
    return country + "_" + name[:4]
```

```python
train1["block_key"] = train1.apply(make_block_key, axis=1)
train3["block_key"] = train3.apply(make_block_key, axis=1)
```

```python
s3_blocks = {}

for idx, row in train3.iterrows():
    key = row["block_key"]
    s3_blocks.setdefault(key, []).append(row["entity_id"])
```

```python
candidate_pairs = []

for _, row in train1.iterrows():
    s1_id = row["entity_id"]
    key = row["block_key"]
    
    
    for candidate_id in s3_blocks.get(key, []):
        candidate_pairs.append(
            (s1_id, candidate_id)
        )
```

```python
candidate_df = pd.DataFrame(
    candidate_pairs,
    columns=["source1_entity_id", "candidate_entity_id"]
)

print(candidate_df.head())
print("Candidate pairs:", len(candidate_df))
```
